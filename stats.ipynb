# **Task 49**

**Machine Learning - Mean Median Mode**

Calculate central tendency measures using NumPy and SciPy.

```python
import numpy
from scipy import stats

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

mean   = numpy.mean(speed)
median = numpy.median(speed)
mode   = stats.mode(speed)

print("Mean:  ", mean)     # 89.76...
print("Median:", median)   # 87.0
print("Mode:  ", mode)     # ModeResult(mode=86, count=3)

# Mean of different dataset
speed2 = [99,86,87,88,86,103,87,94,78,77,85,86]
print("Mean2: ", numpy.mean(speed2))
print("Median2:", numpy.median(speed2))

# Mode example
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
x = stats.mode(ages)
print("Mode of ages:", x)
```

# **Task 50**

**Measure spread of data using std and variance.**

```python
import numpy

speed = [86,87,88,86,87,85,86]
x = numpy.std(speed)
print("Std Dev (low spread):", x)   # ~0.9

speed2 = [32,111,138,28,59,77,97]
y = numpy.std(speed2)
print("Std Dev (high spread):", y)  # ~37.8

# Variance = std squared
var = numpy.var(speed)
print("Variance:", var)

var2 = numpy.var(speed2)
print("Variance2:", var2)

# Percentiles
ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
x = numpy.percentile(ages, 75)
print("75th percentile:", x)

# Multiple percentiles
for p in [25, 50, 75]:
    print(f"{p}th percentile:", numpy.percentile(ages, p))
```
