# **Task 43**

**1-D labeled array (Pandas Series)**

```python
import pandas as pd

# Create a Series from a list
a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)
print(myvar[0])  # access by index

# Custom labels (index)
myvar = pd.Series(a, index=["x", "y", "z"])
print(myvar)
print(myvar["y"])  # access by label

# Create from dictionary
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)

# Only selected keys
myvar = pd.Series(calories, index=["day1", "day2"])
print(myvar)
```

# **Task 44**

**2-D labeled data structure**

```python
import pandas as pd

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data)
print(df)

# Locate row by index
print(df.loc[0])
print(df.loc[[0, 1]])  # multiple rows returns DataFrame

# Named index
df = pd.DataFrame(data, index=["day1", "day2", "day3"])
print(df)
print(df.loc["day2"])
```

# **Task 46**

***Analyze DataFrames with info(), describe(), etc.***

```python
import pandas as pd

df = pd.read_csv('data.csv')

# info() - returns info about the DataFrame
print(df.info())
# Shows: column names, non-null count, dtype

# Null values
print(df.isnull().sum())

# describe() - statistical summary (numeric columns)
print(df.describe())

# head / tail
print(df.head())
print(df.tail())

# Shape
print(df.shape)   # (rows, columns)

# Column names
print(df.columns.tolist())

# Data types
print(df.dtypes)
```

# **Task 47**

**Handle missing, wrong, and duplicate data**

```python
import pandas as pd

df = pd.read_csv('data.csv')

# Remove rows with NULL values
new_df = df.dropna()
print(new_df.to_string())

# Replace NULL values
df.fillna(130, inplace=True)

# Replace only in specific column
df["Calories"].fillna(130, inplace=True)

# Replace using mean
x = df["Calories"].mean()
df["Calories"].fillna(x, inplace=True)

# Wrong data - replace value
df.loc[7, 'Duration'] = 45

# Remove duplicates
print(df.duplicated())
df.drop_duplicates(inplace=True)

print(df.head())
```

# **Task 48**

**Find relationships between columns**

```python
import pandas as pd

df = pd.read_csv('data.csv')

# corr() method - find correlations
# returns correlation matrix (-1 to 1)
print(df.corr())

# Perfect correlation = 1.0
# Perfect negative = -1.0
# No correlation = 0

# Example interpretation:
# Duration vs Calories: 0.922721 (good positive correlation)
# Duration vs Maxpulse: 0.009403 (no correlation)

# Note: Only numeric columns are included
# Strings are ignored automatically

# Visualize with heatmap
import matplotlib.pyplot as plt
import seaborn as sns

sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```
